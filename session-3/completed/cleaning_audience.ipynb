## 3. Tangani missing values


Aturan praktis:
- **Drop rows** kalau hanya sedikit baris terpengaruh, sisa data masih cukup
- **Fill** / **imputasi** kalau proporsi missing besar, drop malah bikin data terlalu kecil

```python
# TODO: fill kolom rating dengan mean dari kolom rating
mean_rating = df["rating"].mean() # mencari mean dari kolom "rating"
print(mean_rating)
df["rating"] = df["rating"].fillna(mean_rating)
```

```python
# TODO: drop baris tanpa city — pakai df.dropna(subset=["city"])
print("Jumlah baris sebelum drop city", len(df))
df = df.dropna(subset=["city"])
print("Jumlah baris setelah drop city", len(df))
```

```python
# TODO: fill product_review yang null dengan string kosong ""
# df[df["product_review"].isna()]
df["product_review"] = df["product_review"].fillna("") 
```

## 4. Hapus duplicates


```python
# TODO: cetak df.duplicated().sum()
print("Jumlah data duplikat", df.duplicated().sum())
```

```python
# TODO: df = df.drop_duplicates()  — lalu cetak len(df)
df = df.drop_duplicates(keep="first")
print("Jumlah data duplikat", df.duplicated().sum())
```

## 5. Standarkan format


```python
# TODO: cetak df["category"].value_counts(). Apa yang bisa kita amati di sini ?
df["category"].value_counts()
```

```python
# TODO: df["category"] = df["category"].str.lower().str.strip()
df["category"] = df["category"].str.lower().str.strip() # merubah kolom kategori menjadi huruf kecil semua, hapus whitespacenya
# Cetak value_counts lagi untuk verifikasi
df["category"].value_counts()
```

```python
# TODO: konversi order_date
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce", format="mixed")
print(df['order_date'].isna().sum())
```

## 6. Bonus — buang noisy values (harga ≤ 0, quantity aneh, rating di luar 0–5)


```python
# TODO: buat boolean mask untuk baris invalid, lalu filter
mask_bad = (
    (df["price_idr"] <= 0) | (df["quantity"] <= 0) | (df["quantity"] > 100) | (df["rating"] < 0) | (df["rating"] > 5)
)
df = df[~mask_bad].copy() # ~ tilde
```
