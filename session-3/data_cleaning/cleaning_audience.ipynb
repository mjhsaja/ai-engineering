## 1. Setup + load dataset


```python
import pandas as pd

# TODO: load orders_dirty.csv ke df
df = pd.read_csv("../datasets/orders_dirty.csv", sep=";", decimal=",")
print("Shape awal:", df.shape)
df.head()
```

## 3. Tangani missing values


```python
# TODO: fill kolom rating dengan mean
mean_rating = df["rating"].mean()
# print(mean_rating)
df["rating"] = df["rating"].fillna(mean_rating)
df.isna().sum()
```

```python
# TODO: drop baris tanpa city — pakai df.dropna(subset=["city"])
df = df.dropna(subset=["city"])
df.isna().sum()
```

```python
# TODO: fill product_review yang null dengan string kosong ""
df["product_review"] = df["product_review"].fillna("")
df.isna().sum()
```

## 4. Hapus duplicates


```python
# TODO: cetak df.duplicated().sum()
print("Duplicate data : ", df.duplicated().sum())
```

```python
# TODO: df = df.drop_duplicates()  — lalu cetak len(df)
before = len(df)
df = df.drop_duplicates()
print(f"Drop {before} baris duplikat. Sisa: {len(df)} baris")
```

## 5. Standarkan format


```python
# TODO: cetak df["category"].value_counts(). Apa yang bisa kita amati di sini ?
print("Jumlah katerogi sebelum dilakukan penyeragaman penulisan")
df["category"].value_counts()
```

```python
# TODO: df["category"] = df["category"].str.lower().str.strip()
# Cetak value_counts lagi untuk verifikasi
df["category"] = df["category"].str.lower().str.strip()
df["category"].value_counts()
```

```python
# TODO: konversi order_date
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce", format="mixed")
df[["order_id", "order_date"]].head(10)
```

## 6. Bonus — buang noisy values (harga ≤ 0, quantity aneh, rating di luar 0–5)


```python
# TODO: buat boolean mask untuk baris invalid, lalu filter
mask_bad = (
    (df["price_idr"] <= 0)                               # harga nol atau negatif
    | (df["quantity"] <= 0) | (df["quantity"] > 100)     # quantity aneh
    | (df["rating"] < 0) | (df["rating"] > 5)            # rating di luar 0–5
)

print("Baris invalid:", mask_bad.sum())
print(df[mask_bad][["order_id", "price_idr", "quantity", "rating"]])

df = df[~mask_bad].copy()
print("Shape setelah filter:", df.shape)
```

```python
# TODO: print shape akhir, total null, kategori unik
print("Shape (jumlah baris dan kolom) akhir : ", df.shape)
print("Total baris yang ada null:", df.isna().sum().sum())
print("Kategori unik:", df["category"].nunique())
```
